# 1. Introducción

## 1.1 Contexto y objetivo del análisis

El presente trabajo tiene como objetivo realizar un **análisis exploratorio de datos (EDA)** para estudiar la relación entre el **Producto Interior Bruto (PIB)** y distintos **indicadores de calidad de vida** en las **comunidades autónomas de España**, a lo largo de un período de varios años.

El crecimiento económico suele medirse mediante indicadores agregados como el PIB; sin embargo, este no siempre refleja de forma directa el bienestar de la población. Por este motivo, resulta de interés analizar cómo evoluciona el PIB en relación con variables que aproximan la calidad de vida, y si dichas relaciones se comportan de forma homogénea o desigual entre las distintas comunidades autónomas y a lo largo del tiempo.

Este análisis tiene un enfoque **descriptivo y exploratorio**, orientado a identificar patrones, tendencias y posibles asociaciones entre variables, sin pretender establecer relaciones causales.

---

## 1.2 Origen de los datos

Los datos utilizados en este trabajo proceden de **fuentes públicas y oficiales**, lo que garantiza su fiabilidad y reproducibilidad. En particular, una parte relevante del conjunto de datos fue obtenida a partir del **Instituto Nacional de Estadística (INE)** de España, a través de su portal web oficial:

> https://www.ine.es

Desde esta plataforma se descargaron indicadores económicos y sociales correspondientes a las comunidades autónomas y a distintos años, en formato estructurado, para su posterior procesamiento y análisis.

---

## 1.3 Construcción del conjunto de datos

El conjunto de datos final utilizado en el análisis **no proviene de una única fuente**, sino que fue construido a partir de **varios datasets obtenidos por separado**, correspondientes a diferentes indicadores y períodos temporales.

El proceso seguido incluyó:
- Selección de variables relevantes para el análisis.
- Homogeneización de nombres, formatos y unidades.
- Integración de los distintos conjuntos de datos mediante procesos de combinación (merge), utilizando como claves comunes la comunidad autónoma y el año.
- Revisión de la consistencia y estructura final del dataset resultante.

Este procedimiento permitió construir un dataset unificado que facilita el análisis conjunto del PIB y de los indicadores de calidad de vida a lo largo del tiempo.

# 02. Hipótesis

## 2.1 Contexto y motivación

El crecimiento económico suele considerarse un factor clave para la mejora del bienestar de la población. Sin embargo, un aumento del Producto Interior Bruto (PIB) no garantiza necesariamente una mejora homogénea de la calidad de vida, ya que dicho crecimiento puede coexistir con desigualdades sociales, dificultades económicas persistentes en determinados grupos poblacionales o problemas estructurales en el acceso a vivienda y recursos básicos.

En el caso de España, las comunidades autónomas presentan trayectorias económicas y sociales heterogéneas, lo que hace relevante analizar si el crecimiento económico se traduce efectivamente en mejoras en distintos indicadores de bienestar y si esta relación se comporta de manera similar entre territorios.

Este análisis se apoya en un conjunto de datos longitudinal que recoge información anual por comunidad autónoma, permitiendo estudiar la evolución temporal de indicadores económicos y sociales y explorar sus asociaciones a lo largo del tiempo.

---

## 2.2 Definición operativa de calidad de vida

La calidad de vida es un concepto complejo y multidimensional que no puede ser capturado por una única variable. En este estudio, se aborda de forma indirecta mediante un conjunto de indicadores socioeconómicos que reflejan distintas dimensiones del bienestar material y social de la población.

En particular, se consideran como aproximaciones a la calidad de vida los siguientes indicadores:

- **Capacidad económica de los hogares**:
  - Porcentaje de población con dificultad para llegar a fin de mes.
  - Porcentaje de población que no puede afrontar gastos imprevistos.

- **Distribución de la riqueza y vulnerabilidad social**:
  - Índice de desigualdad de ingresos (S80/S20).
  - Riesgo de pobreza.

- **Condiciones de vida**:
  - Indicadores relacionados con vivienda y carga económica asociada.

Desde esta perspectiva, un aumento del PIB per cápita no implica necesariamente una mejora global de la calidad de vida si dicho crecimiento no se acompaña de una reducción de las dificultades económicas, una mejora en la capacidad de los hogares para afrontar imprevistos o una disminución de la desigualdad social.

---

## 2.3 Objetivo del análisis

### Objetivo general

Analizar la relación entre el PIB per cápita y distintos indicadores de calidad de vida en las comunidades autónomas de España a lo largo del tiempo, evaluando si el crecimiento económico se asocia con mejoras en el bienestar de la población y si dicha relación es homogénea entre territorios.

---

## 2.4 Formulación de hipótesis

El análisis se guía por un conjunto reducido de hipótesis claras, medibles y verificables, formuladas en términos estadísticos.

### Hipótesis 1: PIB per cápita y dificultad para llegar a fin de mes

- **Hipótesis nula (H₀₁):**  
  No existe correlación entre el PIB per cápita y el porcentaje de población con dificultad para llegar a fin de mes en una comunidad autónoma a lo largo del tiempo.

- **Hipótesis alternativa (H₁₁):**  
  Existe una correlación negativa entre el PIB per cápita y la dificultad para llegar a fin de mes, de modo que mayores niveles de PIB per cápita se asocian con menores dificultades económicas en los hogares.

---

### Hipótesis 2: PIB per cápita y desigualdad de ingresos

- **Hipótesis nula (H₀₂):**  
  No existe correlación entre el PIB per cápita y el nivel de desigualdad de ingresos (S80/S20) en las comunidades autónomas a lo largo del tiempo.

- **Hipótesis alternativa (H₁₂):**  
  Existe una relación no necesariamente negativa entre el PIB per cápita y el nivel de desigualdad de ingresos, observándose patrones heterogéneos entre comunidades autónomas.

---

### Hipótesis 3 : PIB per cápita y capacidad para afrontar gastos imprevistos

- **Hipótesis nula (H₀₃):**  
  No existe correlación entre el PIB per cápita y la capacidad de los hogares para afrontar gastos imprevistos.

- **Hipótesis alternativa (H₁₃):**  
  Un mayor PIB per cápita se asocia con una mayor capacidad de los hogares para afrontar gastos imprevistos.

---
### Hipótesis 4: Dificultad económica y criminalidad

- **Hipótesis nula (H₀₄):**  
  No existe correlación entre el porcentaje de población con dificultad para llegar a fin de mes y la tasa de criminalidad en las comunidades autónomas a lo largo del tiempo.

- **Hipótesis alternativa (H₁₄):**  
  Existe una correlación positiva entre la dificultad para llegar a fin de mes y la tasa de criminalidad, de modo que un aumento de las dificultades económicas se asocia con mayores niveles de criminalidad.
---
### Hipótesis 5: Desigualdad de ingresos y criminalidad

- **Hipótesis nula (H₀₅):**  
  No existe correlación entre el nivel de desigualdad de ingresos (S80/S20) y la tasa de criminalidad en las comunidades autónomas.

- **Hipótesis alternativa (H₁₅):**  
  Mayores niveles de desigualdad de ingresos se asocian con mayores tasas de criminalidad.
---
### Hipótesis 6 : Heterogeneidad territorial

- **Hipótesis nula (H₀₆):**  
  La asociación entre las variables de calidad de vida analizadas es homogénea entre comunidades autónomas.

- **Hipótesis alternativa (H₁₆):**  
  La asociación entre las variables de calidad de vida analizadas varía entre comunidades autónomas.
---

## 2.5 Enfoque metodológico

Dado el carácter longitudinal del conjunto de datos, el análisis se realiza a nivel intra-comunidad, considerando los años disponibles como observaciones. Las asociaciones entre variables económicas y sociales se evalúan mediante el coeficiente de correlación de Spearman, contrastando la hipótesis nula de ausencia de correlación.

Los resultados se interpretan con un enfoque exploratorio y se complementan con análisis gráficos, incluyendo la evolución temporal de las variables y gráficos de dispersión, con el objetivo de facilitar la comprensión de los patrones observados y de la heterogeneidad territorial.


## 3 Análisis exploratorio

### 3.1 Hipotesis 1

**Relación entre el PIB per cápita y la dificultad para llegar a fin de mes**

La Hipótesis 1 plantea la existencia de una relación entre el nivel de actividad económica, medido a través del PIB per cápita, y la proporción de población que manifiesta dificultad para llegar a fin de mes en las comunidades autónomas de España.

---

### 3.1.1 Evolución del PIB per cápita por comunidad autónoma

El análisis de la evolución temporal del PIB per cápita muestra diferencias claras en los niveles absolutos entre comunidades autónomas. No obstante, pese a estas diferencias estructurales, la mayoría de las comunidades presentan **trayectorias temporales similares**, caracterizadas por:

- un descenso del PIB per cápita durante los años de crisis económica,
- una fase posterior de recuperación progresiva,
- y una caída puntual en torno al año 2020, seguida de una recuperación en los años siguientes.

Este comportamiento sugiere la existencia de **factores macroeconómicos comunes** que afectan de forma simultánea a la mayoría de los territorios, aunque partan de niveles económicos distintos.

---

### 3.1.2 Evolución de la dificultad para llegar a fin de mes

En contraste con la evolución del PIB per cápita, la dificultad para llegar a fin de mes presenta **patrones mucho más heterogéneos entre comunidades autónomas**. Se observan diferencias notables tanto en los niveles como en la variabilidad temporal de este indicador.

Algunas comunidades mantienen porcentajes persistentemente elevados de población con dificultades económicas, mientras que otras muestran valores más bajos o trayectorias más estables. Asimismo, la respuesta de este indicador frente a los distintos ciclos económicos no es uniforme, lo que indica que **el impacto del contexto macroeconómico sobre la situación económica de los hogares difiere significativamente entre territorios**.

---

### 3.1.3 Análisis de correlación entre PIB per cápita y dificultad para llegar a fin de mes

Para evaluar la asociación entre el PIB per cápita y la dificultad para llegar a fin de mes, se calculó el coeficiente de correlación de Spearman a nivel intra-comunidad, considerando los años disponibles como observaciones.

En la mayoría de las comunidades autónomas se observa una **correlación negativa estadísticamente significativa** entre ambas variables. Este resultado indica que, en dichos territorios, **mayores niveles de PIB per cápita se asocian con una menor proporción de población con dificultades para llegar a fin de mes**, lo que permite rechazar la hipótesis nula en favor de la hipótesis alternativa.

Sin embargo, en algunas comunidades —entre ellas Illes Balears, Canarias, Cantabria, Melilla y Navarra— no se encontró evidencia estadísticamente significativa de correlación (p ≥ 0,05). En estos casos, no es posible rechazar la hipótesis nula, lo que implica que **no se dispone de evidencia suficiente para afirmar la existencia de una asociación entre ambas variables** en el período analizado.

La ausencia de significación estadística en estas comunidades no implica necesariamente la inexistencia de relación, sino que puede estar asociada a factores como una mayor variabilidad interna, un número limitado de observaciones o características estructurales específicas de sus economías.

---

### 3.1.4 Síntesis de resultados

En conjunto, los resultados indican que, para la mayoría de las comunidades autónomas, existe una asociación negativa entre el PIB per cápita y la dificultad para llegar a fin de mes, lo que respalda la hipótesis de que un mayor nivel de actividad económica se asocia con una menor presión económica sobre los hogares.

No obstante, la presencia de comunidades en las que no se observa una relación estadísticamente significativa pone de manifiesto la existencia de **heterogeneidad territorial**, reforzando la necesidad de analizar las dinámicas económicas y sociales desde una perspectiva regional.
